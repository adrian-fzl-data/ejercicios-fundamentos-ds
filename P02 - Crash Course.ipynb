Crea una función que devuelva el número de veces que aparece la palabra "Data" en un string de entrada de la función.

```python
def count_data(st):
    count = 0
    lista_palabras = st.split()

    for palabra in lista_palabras:
        if palabra == "Data":
            count += 1
            # count = count + 1
    return count
```

```python
count_data("data es la palabra Data que más data aparece en este data texto Data")
```

Para la toma de decisiones de crédito, escribe una función que devuelve tres posibles resultados: "Aprobado", "Duda", "Denegado" según el scoring (de 0 a 100) de un cliente. Se aprueban aquellos con un scoring mayor que 80, van a duda entre 80-50 y se deniega todo por debajo de 50. Salvo que el cliente se llame "Jyts", entonces se aprueba directamente.

```python
def asigna_decision(nombre, scoring):
    if nombre == "Jyts":
        return "Aprobado"
    elif scoring >= 80:
        return "Aprobado"
    elif scoring >= 50:
        return "Duda"
    else:
        return "Denegado"
```

```python
asigna_decision(nombre = "Jyts", scoring = 90)
```

# 📝 Práctica M02: Análisis de Cartera de Microcréditos

**Contexto:** Tenemos una lista de clientes representados como diccionarios. Algunos datos vienen "sucios" (strings con espacios, formatos inconsistentes) y necesitamos procesarlos para decidir a quiénes aprobar un nuevo crédito.   

```python
# DATASET:
cartera_clientes = [
    {'id': 'C-101', 'nombre': '  ALBERTO GARCIA ', 'ingresos': 3000, 'cuotas_impagadas': 0},
    {'id': 'c-102', 'nombre': 'marta sánchez', 'ingresos': 1200, 'cuotas_impagadas': 2},
    {'id': 'C-103', 'nombre': ' LUCIA MARTIN', 'ingresos': 4500, 'cuotas_impagadas': 0},
    {'id': 'C-104', 'nombre': 'pedro gomez ', 'ingresos': 800, 'cuotas_impagadas': 1},
    {'id': 'c-105', 'nombre': 'ANTONIO LOPEZ', 'ingresos': 2200, 'cuotas_impagadas': 0}
]
```

## 🛠️ EJERCICIOS

**Ejercicio 1: Limpieza de nombres**

Usa un bucle for o list comprehension junto con métodos de strings para crear una nueva lista llamada nombres_limpios. Todos los nombres deben estar en mayúsculas y sin espacios sobrantes al principio o al final.

```python
nombres_limpios=[cliente['nombre'].upper().strip() for cliente in cartera_clientes]
print(nombres_limpios)
```

```python
nombres_limpios=[]
for cliente in cartera_clientes:
    nombre=cliente['nombre'] 
    nombres_limpios.append(nombre.upper().strip())
print(nombres_limpios)
```

**Ejercicio 2: Identificación de clientes de riesgo**

Usa la función filter y una lambda expression para obtener una lista de los clientes que tienen 1 o más cuotas_impagadas.

```python
clientes_riesgo=list(filter(lambda cliente: cliente['cuotas_impagadas'] >=1, cartera_clientes))

print("\nClientes con riesgo de impago:")
for c in clientes_riesgo:
    print(f"{c['id']} - Impagos: {c['cuotas_impagadas']}")
```

**Ejercicio 3: Cálculo de capacidad de pago**

Define una función llamada evaluar_credito que reciba los ingresos de un cliente:
* Si los ingresos son mayores a 2000, devuelve "Pre-aprobado".
* Si están entre 1000 y 2000, devuelve "Estudio Manual".
* Si son menores a 1000, devuelve "Rechazado".

No tienes que utilizar el dataset, solo queremos que dados unos ingresos de entrada en la función nos devuelva el texto correspondiente.

```python
def evaluar_credito(ingresos: float):
    '''
    Función que evalúa la capacidad de pago de un cliente en base a sus ingresos 
    '''
    if ingresos>2000:
        return "Pre-Aprobado"
    elif ingresos>=1000:
        return "Estudio Manual"
    else:
        return "Rechazado"

print(f"\nPrueba Individual para un ingreso de 1250: Evaluación Final '{evaluar_credito(1250)}'")
```

```python
evaluar_credito(400)
```


**Ejercicio 4: Transformación masiva**

Usa map para aplicar la función del Ejercicio 3 a todos los ingresos de la cartera y guarda el resultado en una lista.

```python
evaluacion_cartera=list(map(lambda cliente: evaluar_credito(cliente['ingresos']), cartera_clientes))
print(evaluacion_cartera)
```

```python
lista_ingresos=[cliente['ingresos'] for cliente in cartera_clientes]

decisiones_finales=list(map(evaluar_credito, lista_ingresos))
#print("\nDecisiones Finales sobre la Cartera de Clientes:")
#print(decisiones_finales)

for i in range(len(cartera_clientes)):
    id_cliente=cartera_clientes[i]['id'].upper()
    print(f"Cliente {id_cliente} - Decisión Final: {decisiones_finales[i]}")
```

## ✅ SOLUCIONES

```python
# SOLUCIÓN EJERCICIO 1: Limpieza de nombres
# Usamos .strip() para espacios y .upper() para mayúsculas
nombres_limpios = [cliente['nombre'].strip().upper() for cliente in cartera_clientes]

print("Nombres limpios:")
print(nombres_limpios)
```

```python
# SOLUCIÓN EJERCICIO 2: Clientes de riesgo
# Filtramos aquellos donde la clave 'cuotas_impagadas' sea > 0
clientes_riesgo = list(filter(lambda c: c['cuotas_impagadas'] > 0, cartera_clientes))

print("\nClientes con impagos detectados:")
for c in clientes_riesgo:
    print(f"ID: {c['id']} - Impagos: {c['cuotas_impagadas']}")
```

```python
# SOLUCIÓN EJERCICIO 3: Función de evaluación
def evaluar_credito(ingresos):
    """
    Función para determinar el segmento de aprobación basado en ingresos.
    """
    if ingresos > 2000:
        return "Pre-aprobado"
    elif ingresos >= 1000:
        return "Estudio Manual"
    else:
        return "Rechazado"

# Prueba individual
print(f"\nPrueba ingreso 1500: {evaluar_credito(1500)}")
```

```python
# SOLUCIÓN EJERCICIO 4: Transformación masiva con Map
# Extraemos solo los ingresos para procesarlos
lista_ingresos = [c['ingresos'] for c in cartera_clientes]

# Aplicamos la función a toda la lista
decisiones_finales = list(map(evaluar_credito, lista_ingresos))

print("\nDecisiones finales de la cartera:")
print(decisiones_finales)

# BONUS: Verlo de forma elegante
for i in range(len(cartera_clientes)):
    id_c = cartera_clientes[i]['id'].upper()
    print(f"Cliente {id_c}: {decisiones_finales[i]}")
```
