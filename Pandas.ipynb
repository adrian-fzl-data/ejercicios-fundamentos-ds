**Escenario:** Has recibido tres fuentes de datos distintas sobre una pequeña cartera de préstamos. Los datos están desordenados, faltan valores y necesitas consolidarlos para presentar un informe de riesgo a tu responsable.

**Ejercicio 0:** Preparación de los Datos

Ejecuta la siguiente celda para cargar los DataFrames base con los que trabajarás.

```python
import pandas as pd
import numpy as np

# Datos de clientes
clientes_master = pd.DataFrame({
    'ID_Cliente': [f'C-{str(i).zfill(3)}' for i in range(1, 21)],
    'Nombre': [
        'Juan Perez', 'Ana Diaz', 'Luis Ruiz', 'Marta Soto', 'Carlos Gil',
        'Elena Sanz', 'Pedro Gomez', 'Lucia Fernandez', 'Jorge Lopez', 'Maria Garcia',
        'Raul Castro', 'Sofia Martin', 'Andres Molina', 'Paula Vazquez', 'Hugo Ramos',
        'Isabel Torres', 'Miguel Angel', 'Carmen Ortiz', 'Javier Cano', 'Beatriz Rubio'
    ],
    'Ingresos': [3500, np.nan, 4200, 2800, np.nan, 3100, 2900, 5000, 4500, np.nan, 
                 3300, 4100, 2700, 2400, 6000, 3800, 3200, 2100, 4800, 1200],
    'Sucursal': ['Norte', 'Sur', 'Norte', 'Este', 'Sur', 'Norte', 'Sur', 'Este', 'Oeste', 'Norte',
                  'Sur', 'Norte', 'Este', 'Oeste', 'Sur', 'Norte', 'Este', 'Oeste', 'Sur', 'Norte'],
    'Empleo': ['Analista', 'Gerente', 'Analista', 'Cajero', 'Director', 'Analista', 'Cajero', 'Consultor', 'Gerente', 'Analista', 
               'Desarrollador', 'Gerente', 'Analista', 'Soporte', 'Director', 'Consultor', 'Analista', 'Vendedor', 'Gerente', 'Becario'],
    'Edad': [50, 36, 29, 42, 40, 28, 44, 30, 55, 33, 41, 25, 47, 39, 52, 27, 31, 45, 34, 23],
    'Ciudad': ['Madrid', 'Barcelona', 'Madrid', 'Sevilla', 'Valencia', 'Madrid', 'Bilbao', 'Valencia', 'Madrid', 'Zaragoza',
                'Malaga', 'Sevilla', 'Madrid', 'Alicante', 'Murcia', 'Gijon', 'Madrid', 'Santiago', 'Granada', 'Toledo']
})

# Datos del Préstamo
prestamos_data = pd.DataFrame({
    'ID_Cliente': [f'C-{str(i).zfill(3)}' for i in range(1, 21)],
    'Saldo_Pendiente': [120000, 85000, 250000, 10000, 95000, 45000, 130000, 210000, 0, 55000,
                        88000, 15000, 110000, 67000, 290000, 12000, 43000, 0, 180000, 5000],
    'Cuotas_Impagadas': [0, 1, 0, 3, 0, 0, 1, 0, 0, 2, 0, 0, 1, 0, 0, 0, 1, 0, 2, 0]
})
clientes_master.head()
```

## 🛠️ EJERCICIOS

**Ejercicio 1:** Consolidación (Merge)

Crea un nuevo DataFrame llamado df_cartera uniendo clientes_master y prestamos_data a través de su columna común.

```python
df_cartera = pd.merge(clientes_master, prestamos_data, on = 'ID_Cliente' )
df_cartera
#Primero buscamos cual es su columna común, 
#observamos que es el ID del cliente y usar eso como parámetro 'on'
```

**Ejercicio 2:** Tratamiento de Nulos (Fillna)

Existen ingresos faltantes en la cartera.
1. Calcula la mediana de los ingresos de la columna Ingresos.
2. Rellena los valores nulos del df_cartera con esa mediana. No queremos borrar clientes, queremos estimar su capacidad.

```python
mediana_ingrs = df_cartera['Ingresos'].median()
#Simple uso de la función .median() para calcular la mediana de la columna pedida.
df_cartera['Ingresos'] = df_cartera['Ingresos'].fillna(mediana_ingrs)
# Reasignamos la columna con los valores rellenados

#Otra forma válida (se desaconseja cada vez más el uso de inplace) :
df_cartera.fillna({'Ingresos': mediana_ingrs}, inplace = True)
#Con fillna indicamos que vamos a rellenar nulos en df_cartera
#y le indicaremos a la función que será en la columna ingresos con el 
#valor calculado en el parámetro de la mediana
#inplace true para que se formalice en la tabla original este cambio
```

**Ejercicio 3:** Feature Engineering (Nuevas Métricas)

Crea una columna llamada Nivel_Riesgo.
1. Si un cliente tiene Cuotas_Impagadas > 0, su nivel de riesgo es "Alto".
2. De lo contrario, es "Bajo". (Pista: Puedes usar .apply() con una función lambda o np.where()).

```python
df_cartera['Nivel_Riesgo']= np.where(df_cartera['Cuotas_Impagadas']>0, 'Alto', 'Bajo')
#where recibe trs parámetros: 1.Condición a evaluar 2.Valor si verdadero 3.Valor si falso
#Creamos un condicional dentro del where donde le damos los dos valores posibles
#Python recorre toda la columna comprobando la condición y asigna valores
df_cartera.head() 
```

```python
Calculo_Riesgo= lambda Impagos: 'Alto' if Impagos > 0 else 'Bajo'
df_cartera['Nivel_Riesgo']= df_cartera['Cuotas_Impagadas'].apply(Calculo_Riesgo)
#Con .apply() aplicamos una función matemática personalizada o regla de negocio a cada
#celda de una columna en concreto. 
#Creamos una expresión lambda para aplicarla después
#Recorremos la columna aplicando la lambda expression a cada valor
```

```python
df_cartera['Nivel_Riesgo'] = 'Bajo'
df_cartera.loc[df_cartera['Cuotas_Impagadas'] > 0, 'Nivel_Riesgo'] = 'Alto'
#Usamos el indexador bidimensional .loc para decirle que queremos las filas
#que cumplan nuestra condición y la columna del nivel de riesgo
```

**Ejercicio 4:** Recuento

¿Cuál es el empleo que más se repite en nuestra base de datos?

```python
Numero_Empleos=df_cartera.groupby('Empleo')['Empleo'].count()
Empleo_Mas_Repetido = Numero_Empleos.idxmax()
print(f"El empleo más común es: {Empleo_Mas_Repetido}")
```

```python
trabajo_comun = df_cartera['Empleo'].value_counts().idxmax()
#value_counts nos ahora hacer groupby('x')['x'].count() pues hace todo esto
#en una sola instrucción, devolviendo una serie de pandas a la cual 
#le aplicamos idxmax() para que nos de cual es
#el indice que mas se repite y lo sacamos por pantalla
print(f"El trabajo más común es: {trabajo_comun}")
```

**Ejercicio 5:** Diversidad laboral

¿Cuántos trabajos distintos existen en el dataset?

```python
Trabajos_Dist = df_cartera['Empleo'].nunique()
# .nunique() devuelve el número de objetos diferentes que hay en el objeto
print(f"Hay {Trabajos_Dist} trabajos distintos en nuestro dataset")
```

**Ejercicio 6:** Ciudades únicas

¿Qué ciudades aparecen exactamente una sola vez?

```python
Conteos_Ciudad = df_cartera['Ciudad'].value_counts()
#Sacamos una pd.serie de el número de clientes que tiene cada ciudad
Ciudades_Unicas = Conteos_Ciudad[Conteos_Ciudad==1].index.tolist()
#Buscamos las que solo tienen un único cliente y sacamos los índices de este df
#y lo pasamos a una lista que devolvemos por pantalla
print(f"Las ciudades que aparecen de forma única son: {Ciudades_Unicas}")
```

**Ejercicio 7:** Búsqueda por nombre

Identifica a todas las personas que tengan la letra "u" (mayúscula o minúscula) en su nombre.

```python
contiene_u = df_cartera[df_cartera['Nombre'].str.contains("u", case = False)]
#Usamos .str para que la pd.series devuelta por la selección sea tratada como un string
#y así podamos aplicarle la función .contains a la cual le indicaremos 
#que no sea case sensitive en este caso específico
print("Clientes con 'u' en su nombre: ")
print(contiene_u['Nombre'])
#Filtramos de la tabla de los clientes para obtener solo su nombre y 
#no el resto de su información irrelevante. 
```

**Ejercicio 8:** Correlación

Calcula la correlación entre la Edad y los Ingresos. ¿Existe alguna relación lineal?

```python
correlacion = df_cartera['Edad'].corr(df_cartera['Ingresos'])
#De este modo calcula una correlacion unica, lo que devuelve un número,
#para calcular otras podemos hacer
# 1.) Entre varios parámetros entre si, lo que devuelve un df (matriz correlación): 
#     df_cartera[['Edad','Ingresos','Cuotas_Impagadas']].corr()

# 2.) Un parámetro frente al resto del df, lo que devuelve una pd.series: 
#       1. df_cartera.corr()['Cuotas_Impagadas'] 
#       2. df_cartera.corrwith(df_cartera['Cuotas_Impagadas'])
print(f"La correlación entre edad e ingresos es: {correlacion:.2f}")
```

**Ejercicio 9:** Análisis Agregado (GroupBy)

¿Cuál es el Saldo_Pendiente total y el Ingreso promedio por cada Sucursal? Presenta una tabla resumen.

```python
df_cartera.groupby('Sucursal').agg({
    'Saldo_Pendiente' : 'sum',
    'Ingresos' : 'mean'
})
#Primero agrupamos por sucursales que es lo que nos piden,
#después usamos .agg() que acepta el paso de diccionarios con
#estructura {'Nombre_Col': 'Operacion_Mat'}, además
#podríamos pasarle una lista como operaciones para una columna.
# ... 'Ingresos' : ['mean','max']... })
```

**Ejercicio 10:** Selección Crítica

Muestra únicamente los datos de los clientes de la sucursal "Sur" que tengan un Saldo_Pendiente mayor a 90.000€.

```python
Sucursal_Sur_High = df_cartera[(df_cartera['Sucursal']=='Sur') & (df_cartera['Saldo_Pendiente'] > 90000)]
#Puesto que se nos piden valores centrales dentro de la tabla debemos hacer un filtrado doble
#como si de una intersección en teoría de conjuntos se tratase.
print("\n --- Alerta Clientes Sucursal Sur con alto Saldo Pendiente---")
print(Sucursal_Sur_High)
```

## ✅ SOLUCIONES

```python
# 1. Merge
df = pd.merge(clientes_master, prestamos_data, on='ID_Cliente')

# 2. Imputación de nulos
mediana_ingr = df['Ingresos'].median()
df['Ingresos'].fillna(mediana_ingr, inplace=True)

# 3. Clasificación de riesgo
df['Nivel_Riesgo'] = df['Cuotas_Impagadas'].apply(lambda x: 'Alto' if x > 0 else 'Bajo')

# 4. Trabajo más común
trabajo_comun = df['Empleo'].value_counts().idxmax()
print(f"El trabajo más común es: {trabajo_comun}")

# 5. Cantidad de trabajos distintos
total_empleos = df['Empleo'].nunique()
print(f"Hay {total_empleos} tipos de empleos diferentes.")

# 6. Ciudades que aparecen solo una vez
conteos_ciudades = df['Ciudad'].value_counts()
ciudades_unicas = conteos_ciudades[conteos_ciudades == 1].index.tolist()
print(f"Ciudades con un solo cliente: {ciudades_unicas}")

# 7. Personas con una "u" en su nombre
con_u = df[df['Nombre'].str.contains('u', case=False)]
print("Clientes con 'u' en su nombre:")
print(con_u[['Nombre']])

# 8. Correlación entre Edad e Ingresos
# Nota: corr() ignora los valores NaN automáticamente
correlacion = df['Edad'].corr(df['Ingresos'])
print(f"La correlación entre Edad e Ingresos es: {correlacion:.2f}")

# 9. Agregación por sucursal
resumen_sucursales = df_cartera.groupby('Sucursal').agg({
    'Saldo_Pendiente': 'sum',
    'Ingresos': 'mean'
})

# 10. Filtrado condicional
filtro_sur_high = df_cartera[(df_cartera['Sucursal'] == 'Sur') & (df_cartera['Saldo_Pendiente'] > 90000)]

# Visualización de resultados finales
print("--- Cartera Consolidada ---")
print(df_cartera)
print("\n--- Resumen por Sucursal ---")
print(resumen_sucursales)
print("\n--- Alerta Clientes Sur con Alto Saldo ---")
print(filtro_sur_high)
```
